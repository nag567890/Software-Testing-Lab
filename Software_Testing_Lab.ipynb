```python
# Verification and Validation

## Verification

Checks whether software is developed according to requirements.

Examples:
- Requirement review
- Design review
- Code inspection


## Validation

Checks whether software meets user needs.

Examples:
- Running software
- Testing output
```

```python
def check_marks(marks):

    if marks >= 50:
        return "PASS"
    else:
        return "FAIL"


print(check_marks(80))
print(check_marks(30))
```

## Test Cases

|Test Case|Input|Expected Output|

```python
total_statements = 4
executed_statements = 4

coverage = (executed_statements / total_statements)*100

print("Statement Coverage =",coverage,"%")
```

```python
total_branches = 2
executed_branches = 2

coverage=(executed_branches/total_branches)*100

print("Branch Coverage =",coverage,"%")
```

```python
import networkx as nx
import matplotlib.pyplot as plt


G=nx.DiGraph()

G.add_edges_from([
("Start","Decision"),
("Decision","PASS"),
("Decision","FAIL"),
("PASS","End"),
("FAIL","End")
])


nx.draw(G,with_labels=True)

plt.show()
```

```python
nodes=5
edges=5

complexity=edges-nodes+2

print("Cyclomatic Complexity =",complexity)
```
